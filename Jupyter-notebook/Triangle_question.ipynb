<h3>Check if a triangle is valid based on angles</h3>

```python
print("_"*30)
while True:
    try:
        side1 = int(input("ENTER FIRST ANGLE"))
        side2 = int(input("ENTER SECOND ANGLE"))
        side3 = int(input("ENTER THIRD ANGLE"))
        all_side = side1 + side2 + side3
        if all_side == 180:
            print("triangle is valid based on angles you've entered")
            break
        else:
            print("triangle isn't valid")
            break
    except:
        print("invalid input")
        usr_con = input("wanna try again (y/n):: ")
        if usr_con == "y":
            continue
        elif usr_con == "n":
            break
        else: 
            print("invalid input")
            break
```

<h3>Check whether the triangle is scalane, equilateral, isosceles</h3>

```python
#let's take all side length
def invalid_triangle(x,y,z):
    return (x+y>z) and (x+z>y) and (z+y>x)
#___________________________________________________
try:
    sd1 = int(input("Enter 1st SIDE LENGTH:: "))   
    sd2 = int(input("Enter 2nd SIDE LENGTH:: "))
    sd3 = int(input("Enter 3rd SIDE LENGTH:: "))
    if invalid_triangle(sd1,sd2,sd3):
        if sd1 == sd2 == sd3:
            print("equilateral triangle")
        elif (sd1 == sd2 != sd3) or (sd1 == sd3 != sd2):
            print("Isoscele triangle")
        else:
            print("Scalene triangle")
    elif not invalid_triangle(sd1,sd2,sd3):
        print("Triangle sides aren't satisfied according to theorem")
except:
    print("invalid input")
        

```
