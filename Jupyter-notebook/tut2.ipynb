<h3>Weight converter program</h3>

```python
weight = float(input("Enter your weight:: "))
unit = input("Kilogram or Pounds ? (K/L):: ").upper()
if unit == "K":
    weight = weight * 2.205
    print(f" Your weight is: {round(weight,2)} Lbs.")
elif unit == "L":
    weight = weight / 2.205
    print(f" Your weight is: {round(weight,2)} Kgs.")
else:
    print(f" {unit} is not valid unit")
```

<h3>Temperature converter program</h3>

```python
#this program convert temperature unit into another unti like if you enter f it will convert it into c
usr_temp = float(input("Enter TEMPERATURE:: "))
usr_unit = input("Enter your temperature unti in (C/F):: ").upper()
if usr_unit == "C":
    usr_temp = round((9 * usr_temp) / 5 + 32,2)
    print(f"Your temperature in F is:: {usr_temp} deg. F")
elif usr_unit == "F":
    usr_temp = round((usr_temp - 32) * 5/9,2)
    print(f"Your temperature in C is:: {usr_temp} deg. C")
else:
    print(f"{usr_unit} is not valid unit")
```
